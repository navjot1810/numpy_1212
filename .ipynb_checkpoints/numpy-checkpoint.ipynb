# What is a Library?

* A library is a collection of pre-written code that helps programmers perform tasks quickly without writing everything from scratch.

* Real-World Example

    * Imagine you are building a house.

    * Making every brick yourself = Writing all code from scratch
    * Buying ready-made bricks = Using libraries

* Libraries save time and effort.

# Why Do We Use Libraries?
* Benefits
    * Reuse existing code
    * Save development time
    * Reduce errors
    * Increase productivity
    * Access advanced features easily

| `ndim` | Returns the number of dimensions of the array | `arr.ndim` |

# NumPy Indexing

| Type | Description | Example |
|---|---|---|
| Positive Indexing | Access elements using index starting from `0` | `arr[0]` |
| Negative Indexing | Access elements from the end | `arr[-1]` |
| 2D Indexing | Access an element using row and column | `arr[0, 1]` |
| Row Indexing | Access a complete row | `arr[0, :]` |
| Column Indexing | Access a complete column | `arr[:, 0]` |
| Multiple Indexing | Access specific elements using index positions | `arr[[0, 2]]` |

### Array Creation Functions :-

**Array** :- *An array is the main data structure of NumPy. It stores multiple values in a single variable, and all values are of the same data type.*


```python
a = np.array([1, 2, 3, 4])
a
```

**Zeros** = The numpy.zeros() function creates an array filled with 0s.

```python
arr = np.zeros((4,4),dtype=int)
arr
```

**Ones** :- *When we want to initialize an array with the default value 1, we use `ones()`.*


```python
arr = np.ones((6,3))
arr
```

**eye** :- *An identity matrix is a matrix in which the diagonal elements are 1, and all other elements are 0.*


```python
arr = np.eye(4,4,dtype=int)
arr
```

**full** :- *Fills every element of the array with the same value for the given size.*


```python
arr = np.full((4,4), 0)
arr
```

### Mathematical Functions :-

```python
arr = np.array([10, 20, 30, 40,50,70])
print(arr.sum())
```

**sum** :- *Gives the sum of all the values present in the array at once.*


```python
arr = np.array([10, 20, 30, 40,50])
print(arr.mean())
```

**mean** :- *A NumPy function that calculates the average (arithmetic mean) of an array.*
**Formula:** Mean = (Sum of all values) ÷ (Total number of values)


```python
arr = np.array([10, 45, 23, 90, 12])
print(arr.max())
```

**max** :- *Finds the largest (maximum) value in the array.*


```python
arr = np.array([10, 45, 23, 90, 12])
print(arr.min())
```

**min** :- *Finds the smallest (minimum) value in the array.*


```python
arr = np.array([1, 4, 9, 16, 25,36])
print(np.sqrt(arr))
```

**sqrt** :- *Calculates the square root of each element in the array.*


### Shape & Dimension Functions :-

**ndim** :- *Tells the total number of dimensions of the array.*


```python
arr = np.array([10, 20, 30])
print(arr.ndim)
```

**shape** :- *Tells the structure or size of the array (rows, columns, blocks).*


```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr.shape)
```

**size** :- *Tells the total number of elements in the array.*


```python
arr = np.array([10, 20, 30, 40,50,80,7])
print(arr.size)
```

### Random Functions :- (random.rand,random.randint,random.randn)

**random.rand** :- *A NumPy function that generates random numbers between 0 and 1 (positive values).*


```python
arr = np.random.rand(3,3)
print(arr)
```

**random.randint** :- *Generates random integers within a specified range.*
**Example Range:** (10, 100)


```python
arr = np.random.randint(10,100,(4,3))
print(arr)
```

**random.randn** :- *Generates normally distributed (Gaussian) random numbers. These numbers can be both negative and positive, generally around a mean of 0 and standard deviation of 1.*


```python
arr = np.random.randn(11)
print(arr)
```

### Sequence Functions :- 

**arange** :- *Generates numbers from start to stop with a specified step.*


```python
arr = np.arange(0, 40, 2)   # (start, stop, step)
print(arr)
```

**linspace** :- *Generates evenly spaced numbers between a specified start and stop.*


```python
arr = np.linspace(0,20,5)  # (start,stop num)
print(arr)
```

### Matrix Functions :- 

**dot** :- *Performs the dot product (sum of products) or matrix multiplication.*


```python
a = np.array([1, 6, 3])
b = np.array([4, 5, 6])
result = np.dot(a, b)
print(result)
```

**transpose (T)** :- *Swaps the rows and columns of a matrix.*


```python
arr= np.array([[1, 2, 3],[4, 5, 6]])
print(arr.T)
```

**argmax()** :- *A NumPy function that returns the index position of the largest (maximum) value in the array.*


```python
arr = np.array([10, 45, 230, 9, 34])
print(np.argmax(arr))
```

**argmin()** :- *A NumPy function that returns the index position of the smallest (minimum) value in the array.*


```python
arr = np.array([10, 45, 23, 55, 34])
print(np.argmin(arr))
```

**std()** :- *A NumPy function that calculates the standard deviation. It measures how much the data is spread out or deviates from the mean (average).*


```python
arr = np.array([10, 20, 30, 40, 50])
print(np.std(arr))
```

**flatten()** :- *A NumPy function that converts a multi-dimensional array (2D, 3D, etc.) into a 1D array.*


```python
arr = np.array([[1, 2, 3],[4, 5, 6]])
print(arr.flatten())
```

**ravel()** :- *Flattens the array, but in most cases returns a “view” of the original array rather than a copy.*


```python
arr = np.array([[1, 2, 3],[4, 5, 6]])
print(arr.ravel())
```

**reshape()** :- *A NumPy function that changes the shape of an array. The data remains the same; only the rows–columns structure is modified.*


```python
arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.reshape(2, 3))
```

**diag()** :- *A NumPy function that works with the diagonal elements of an array or matrix.*


```python
arr = np.array([1, 2,3,4])
print(np.diag(arr))
```

## Difference Between NumPy Array and Python List

| Feature                  | Python List                                                              | NumPy Array                                                                                          |
| ------------------------ | ------------------------------------------------------------------------ | ---------------------------------------------------------------------------------------------------- |
| **Performance**          | Slower for numerical computations due to per-element overhead            | Much faster for large-scale mathematical operations                                                  |
| **Mathematical Support** | Limited to general-purpose methods like `sort()`, `count()`, `reverse()` | Supports advanced mathematical functions, linear algebra, broadcasting, and multi-dimensional arrays |
| **Data Type**            | Can store mixed data types in a single list                              | All elements must be of the same data type                                                           |
| **Memory Efficiency**    | Less memory efficient                                                    | More memory efficient for numerical data                                                             |
| **Usage**                | General-purpose container                                                | Optimized for scientific and numerical computing                                                     |


*In Python, both lists and NumPy arrays are used to store data. Their functionality may seem similar, but there is a significant difference in performance, usage, and capability.*

*NumPy performs faster for large-scale mathematical operations, while lists are slower for numerical computations due to per-element overhead.*

*NumPy supports advanced mathematical functions, linear algebra, broadcasting, and multi-dimensional arrays, whereas lists offer general-purpose methods like sort, count, and reverse.*


```python
list1 = [1, 2, 3, 4]
array1 = np.array([1,2,3,4])
print("List Addition: ",list1 + list1)
print("Numpy array Addition : ", array1 +array1)
```

*Lists repeat or concatenate the values, whereas NumPy arrays perform element-wise addition and return the sum of values.*
